# Principal Component Analysis (PCA) Tutorial

This notebook provides a detailed explanation of Principal Component Analysis (PCA), a popular dimensionality reduction technique. We will cover the theory behind PCA, its implementation from scratch, and its application using the scikit-learn library.

## 1. Introduction to PCA

PCA is a linear dimensionality reduction technique that can be utilized for extracting information from a high-dimensional space by projecting it into a lower-dimensional sub-space. It tries to preserve the essential parts that have more variation of the data and remove the non-essential parts with fewer variation.

## 2. Data Preparation

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
# Load the Iris dataset
iris = load_iris()
X = iris.data
y = iris.target
```

```python
# Standardize the features
X_scaled = StandardScaler().fit_transform(X)
```

## 3. PCA Implementation from Scratch

### 3.1. Calculate the Covariance Matrix

```python
covariance_matrix = np.cov(X_scaled.T)
```

### 3.2. Compute Eigenvectors and Eigenvalues

```python
eigen_values, eigen_vectors = np.linalg.eig(covariance_matrix)
```

### 3.3. Sort Eigenvectors by Eigenvalues

```python
sorted_index = np.argsort(eigen_values)[::-1]
sorted_eigenvalue = eigen_values[sorted_index]
sorted_eigenvectors = eigen_vectors[:,sorted_index]
```

### 3.4. Select Top k Eigenvectors

```python
n_components = 2
eigenvector_subset = sorted_eigenvectors[:,0:n_components]
```

### 3.5. Transform the Data

```python
X_reduced = np.dot(eigenvector_subset.T,X_scaled.T).T
```

## 4. PCA using Scikit-learn

```python
pca = PCA(n_components=2)
X_reduced_sklearn = pca.fit_transform(X_scaled)
```

## 5. Explained Variance

The explained variance tells you how much of the original variance is captured by each principal component. By plotting the cumulative explained variance, we can decide how many principal components to keep.

```python
pca = PCA().fit(X_scaled)
plt.plot(np.cumsum(pca.explained_variance_ratio_))
plt.xlabel('Number of Components')
plt.ylabel('Cumulative Explained Variance')
plt.title('Explained Variance by Principal Components')
plt.show()
```

## 6. Visualize the Transformed Data

```python
plt.figure(figsize=(8, 6))
scatter = plt.scatter(X_reduced_sklearn[:, 0], X_reduced_sklearn[:, 1], c=y, cmap='viridis')
plt.title('PCA of Iris Dataset')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend(handles=scatter.legend_elements()[0], labels=iris.target_names.tolist())
plt.show()
```
